```python
import sqlite3

conn = sqlite3.connect("college.db")
cursor = conn.cursor()

print("Database Connected Successfully")
cursor.execute("""
CREATE TABLE IF NOT EXISTS Student(
    RollNo INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER,
    Department TEXT
)
""")

print("Table Created Successfully")
cursor.execute("INSERT INTO Student VALUES (101,'Rahul',19,'CSE')")
cursor.execute("INSERT INTO Student VALUES (102,'Priya',20,'ECE')")
cursor.execute("INSERT INTO Student VALUES (103,'Arun',18,'IT')")

conn.commit()

print("Records Inserted Successfully")
cursor.execute("SELECT * FROM Student")

rows = cursor.fetchall()

for row in rows:
    print(row)
cursor.execute("SELECT Name, Department FROM Student")

rows = cursor.fetchall()

for row in rows:
    print(row)
cursor.execute("SELECT * FROM Student WHERE Age > 18")

rows = cursor.fetchall()

for row in rows:
    print(row)
conn.close()

print("Database Closed")
```
