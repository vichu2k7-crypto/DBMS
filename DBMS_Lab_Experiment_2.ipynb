```python
import sqlite3

conn = sqlite3.connect("college.db")
cursor = conn.cursor()

print("Database Connected Successfully")
cursor.execute("""
CREATE TABLE Department(
DeptID INTEGER PRIMARY KEY,
DeptName TEXT NOT NULL UNIQUE
)
""")

print("Department Table Created")
cursor.execute("""
CREATE TABLE Student(
RollNo INTEGER PRIMARY KEY,
Name TEXT NOT NULL,
Age INTEGER CHECK(Age>=18),
Department TEXT DEFAULT 'CSE',
DeptID INTEGER,
FOREIGN KEY(DeptID) REFERENCES Department(DeptID)
)
""")

print("Student Table Created")
cursor.execute("INSERT INTO Department VALUES(1,'CSE')")
cursor.execute("INSERT INTO Department VALUES(2,'ECE')")

conn.commit()

print("Department Data Inserted")
cursor.execute("""
INSERT INTO Student
(RollNo,Name,Age,DeptID)
VALUES
(101,'Rahul',19,1)
""")

conn.commit()

print("Student Data Inserted")
cursor.execute("SELECT * FROM Department")
print(cursor.fetchall())

cursor.execute("SELECT * FROM Student")
print(cursor.fetchall())
conn.close()
```
